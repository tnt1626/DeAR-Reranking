# MedPRS Journal Recommendation and Listwise Reranking Notebook

This notebook demonstrates the two-stage journal recommendation pipeline on Kaggle:
1. **Stage 1 (Pointwise Candidate Selection)**: Uses a fine-tuned BioBERT model (`dmis-lab/biobert-v1.1` + `Epoch_02_SIMCPRS_dmis-lab_biobert-v1_1_CL.pth` checkpoint) to retrieve the top 10 suitable journals from a database of 1,408 journals for a given paper.
2. **Stage 2 (Listwise Reranking)**: Uses the DeAR Llama-3.1-8B Listwise model (`abdoelsayed/dear-8b-reranker-listwise-lora-v1`) to compare the top 10 journals simultaneously inside a custom multi-turn prompt, sorting them based on their suitability to the paper's title and abstract.
